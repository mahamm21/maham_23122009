```python
!pip install -U google-genai
from google import genai

client = genai.Client(api_key="")

MODEL = "models/gemini-3.5-flash"

fuuast_info = {
    "university": "Federal Urdu University (FUUAST)",
    "campus": "Gulshan Campus, Karachi",
    "total_departments": 15,
    "popular_departments": [
        "Computer Science",
        "Software Engineering",
        "Business Administration (BBA)",
        "Pharmacy",
        "Biotechnology",
        "Chemistry",
        "Physics",
        "Mathematics"
    ],
    "semester_fee": "Around PKR 50,000 to 80,000 per semester",
    "admissions": "Twice a year (Fall & Spring)"
}

def ask_agent(question):
    prompt = f"""
You are a friendly student help-desk assistant for FUUAST (Gulshan Campus, Karachi).

Here is the quick information you know:
- University: {fuuast_info['university']}
- Campus: {fuuast_info['campus']}
- Total Departments: {fuuast_info['total_departments']}
- Popular Departments: {', '.join(fuuast_info['popular_departments'])}
- Fees: {fuuast_info['semester_fee']} for all programs
- Admissions: {fuuast_info['admissions']}

Instructions:
1. Answer the student's question in a casual, friendly, and helpful tone (like a senior student helping a junior).
2. Don't make answers too long keep them short and clear.

Student question: "{question}"

Helpful response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text


print("FUUAST Help Desk is live! (Type 'exit' to stop)\n")

while True:
    user_msg = input("You: ")

    if user_msg.strip().lower() == "exit":
        print("Bye, good luck!")
        break

    try:
        reply = ask_agent(user_msg)
        print("\nAssistant:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
