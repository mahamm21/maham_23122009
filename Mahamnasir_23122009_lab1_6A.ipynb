```python
print("Flight Booking System")

departure = input("Enter departure city: ")
destination = input("Enter destination city: ")

if departure == "Karachi" and destination == "Lahore":
    print("F101 - Rs.15000")
    print("F102 - Rs.18000")

    flight = input("Enter flight ID: ")

    if flight == "F101":
        print("Price: Rs.15000")
    elif flight == "F102":
        print("Price: Rs.18000")
    else:
        print("Invalid flight ID")

    if flight == "F101" or flight == "F102":
        confirm = input("Book flight? yes/no: ")

        if confirm == "yes":
            print("Flight booked successfully")
        else:
            print("Booking cancelled")

elif departure == "Lahore" and destination == "Karachi":
    print("F201 - Rs.16000")
    print("F202 - Rs.19000")

    flight = input("Enter flight ID: ")

    if flight == "F201" or flight == "F202":
        print("Flight booked successfully")
    else:
        print("Invalid flight ID")

else:
    print("No flights available")
```
